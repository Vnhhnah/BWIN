```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

from xgboost import XGBClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score, confusion_matrix

# Make sure to mount your Google Drive and verify the file path
path = '/content/drive/MyDrive/BWIN/Data default.csv'
df = pd.read_csv(path)
print(df.isnull().sum())

df.head()
```

```python
import statsmodels.api as sm
from sklearn.metrics import roc_auc_score

# Xử lý giá trị thiếu cho các biến
df['Delay-from-due-date'] = df['Delay-from-due-date'].fillna(df['Delay-from-due-date'].mean())
df['Num-of-Delayed-Payment'] = df['Num-of-Delayed-Payment'].fillna(df['Num-of-Delayed-Payment'].mean())
df['Outstanding-Debt'] = df['Outstanding-Debt'].fillna(df['Outstanding-Debt'].mean())
df['Credit-Utilization-Ratio'] = df['Credit-Utilization-Ratio'].fillna(df['Credit-Utilization-Ratio'].mean())

# Tạo DataFrame X với các biến độc lập
X = df[['Delay-from-due-date', 'Num-of-Delayed-Payment', 'Outstanding-Debt', 'Credit-Utilization-Ratio',
        'Debt-to-Income Ratio', 'Amount-invested-monthly', 'Monthly-Balance', 'Credit-History-Age', 'Num-Credit-Inquiries']]
X = sm.add_constant(X)  # Thêm hằng số vào mô hình

# Biến phụ thuộc y (Default)
y = df['Default']

# Huấn luyện mô hình Logistic Regression
logit_model = sm.Logit(y, X).fit()
print(logit_model.summary())

# Tính ROC-AUC để đánh giá mô hình
roc_auc = roc_auc_score(y, logit_model.predict(X))
print("ROC-AUC:", roc_auc)
```
